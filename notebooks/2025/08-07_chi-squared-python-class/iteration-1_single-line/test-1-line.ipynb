```python
from chi_squared_fitter import ChiSquaredFitter

import matplotlib.pyplot as plt
import numpy as np

# --- Helper function to load a single line from your .dat file ---
def load_dat_star_data(dat_path, index=0):
    with open(dat_path, 'r') as f:
        lines = f.readlines()

    line = lines[index].strip().split()
    line = [float(x) for x in line]

    # Assuming same column format as before:
    f162m, err_f162m = line[2], line[3]
    f182m, err_f182m = line[4], line[5]
    f200w, err_f200w = line[6], line[7]
    av = line[8]
    mass = line[9]

    mags = [f162m, f182m, f200w]
    errs = [err_f162m, err_f182m, err_f200w]
    return mags, errs, av, mass

# --- Load the test data ---
dat_path = '/scratch/wyz5rge/synthetic-hr/notebooks/2025/08-07_chi-squared-python-class/test-file_1.dat'
mags, errs, av_true, mass_true = load_dat_star_data(dat_path, index=0)

# --- Initialize fitter and run analysis ---
fitter = ChiSquaredFitter(output_dir='./fit_plots_test')
result = fitter.analyze_line(index=0, mags=mags, errs=errs, true_av=av_true, true_mass=mass_true)

# --- Print returned fit results ---
print("Fit results for line 0:")
for k, v in result.items():
    print(f"{k}: {v}")

# --- Display the saved plot ---
plot_path = f"./fit_plots_test/fit_0.png"

# Show the plot image inline (requires matplotlib)
img = plt.imread(plot_path)
plt.figure(figsize=(10, 7))
plt.imshow(img)
plt.axis('off')
plt.title("Fit Plot for Line 0")
plt.show()
```
