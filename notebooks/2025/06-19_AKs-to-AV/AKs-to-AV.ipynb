```python
# === Extinction Law: A_Ks / A_V Calculation ===

from spisea import reddening

# Instantiate the Cardelli (1989) extinction law with standard ISM value
Rv = 3.1
red_law = reddening.RedLawCardelli(Rv=Rv)

# Define wavelengths in microns
wv_V = 0.55    # V-band (Johnson-Cousins)
wv_Ks = 2.15   # Ks-band (2MASS-like)

# Get extinction values for A_lambda, assuming AKs = 1
A_V = red_law.Cardelli89(wv_V, AKs=1.0)[0]
A_Ks = red_law.Cardelli89(wv_Ks, AKs=1.0)[0]

# Compute and display the ratios
print(f"Cardelli (1989) extinction law with Rv={Rv}")
print(f"A(V)  = {A_V:.4f} mag")
print(f"A(Ks) = {A_Ks:.4f} mag")
print(f"A(Ks)/A(V) = {A_Ks / A_V:.4f}")
print(f"A(V)/A(Ks) = {A_V / A_Ks:.4f}")
```
